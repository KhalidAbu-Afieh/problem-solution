```python
#!/bin/python3

import math
import os
import random
import re
import sys

#
# Complete the 'Mode' function below.
#
# The function is expected to return an INTEGER.
# The function accepts following parameters:
#  1. INTEGER n
#  2. INTEGER_ARRAY values
#
#n= number of value
#values =  list of integer
def Mode(n, values):
    freq = {}      # create an empty dictionary to store frequency of each number 

    # Count frequency of each value
    for num in values: # Loops through each number in the values list
        if num in freq:# if number already seedn before
            freq[num] += 1 # increase  its count by 1
        else:
            freq[num] = 1 #if not add the number to dictionary with count 1

    # Find maximum frequency
    max_freq = max(freq.values()) #find the highest occurrence count among all the number

    # Count how many numbers have this frequency
    mode_value = -1   # Will store the number that is the mode
    count_modes = 0 # Counts how many numbers share the max frequency
    
  # Loop through each number in the frequency dictionary
    for num in freq:
        if freq[num] == max_freq:  # If this number has the highest frequency
            mode_value = num # Save it as the mode
            count_modes += 1   # Increase count of modes found

            # If more than one mode, return -1
    if count_modes > 1:  # If multiple numbers share the same highest frequency
        return -1      # Return -1 because distribution is multimodal
    else:
        return mode_value  # Otherwise return the single mode value


if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    n = int(input().strip())

    values = []

    for _ in range(n):
        values_item = int(input().strip())
        values.append(values_item)

    mode = Mode(n, values)

    fptr.write(str(mode) + '\n')

    fptr.close()
```

```python
#!/bin/python3

import math
import os
import random
import re
import sys

#
# Complete the 'UniquePairs' function below.
#
# The function is expected to return an INTEGER.
# The function accepts following parameters:
#  1. INTEGER n
#  2. INTEGER_ARRAY values
#
def UniquePairs(n, values):
     # Convert list to a SET to remove duplicates
    # len(set(values)) gives how many different numbers exist
    
    unique_count = len(set(values))

    # Calculate number of unordered pairs
    # pairs =  k * (k - 1) // 2
    
    pairs = unique_count * (unique_count - 1) // 2

    # Return the total number of unique pairs
    return pairs



if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    n = int(input().strip())

    values = []

    for _ in range(n):
        values_item = int(input().strip())
        values.append(values_item)

    count = UniquePairs(n, values)

    fptr.write(str(count) + '\n')

    fptr.close()
```

```python
#!/bin/python3

import math
import os
import random
import re
import sys

#
# Complete the 'ParkingTickets' function below.
#
# The function is expected to return an INTEGER.
# The function accepts following parameters:
#  1. INTEGER numEvents
#  2. STRING_ARRAY registrations
#  3. INTEGER_ARRAY times
def ParkingTickets(numEvents, registrations, times):
    # Create a dictionary to store information about each car
    # Key is car registration number and Value is dictionary of that car's detail
    car_info = {}
    
    #  the number of tickets issued to 0
    tickets = 0
    
    # Loop over each car registration and its corresponding event time
    # zip() pairs each registration with its corresponding time
    for reg, t in zip(registrations, times):
        
        # If this car is seen for the first time create a new entry for it
        if reg not in car_info:
            car_info[reg] = {
                "last_exit": None,      # Time when the car last left the parking
                "ticketed": False,      # Has the car been given a ticket already
                "currently_parked": False,  # Is the car in the parking lot right now
                "enter_time": None      # Time when the car entered the parking lot
            }
        
        # Get the dictionary of this specific car for easier access
        car = car_info[reg]
        
        # Check if the car is currently not in the parking lot
        if not car["currently_parked"]:
            # The car is entering the parking lot
            
            # If the car has exited before, check the gap between last exit and this entry
            if car["last_exit"] is not None:
                gap = t - car["last_exit"]  # Time difference between last exit and this entry
                
                # If the gap is less than 60 minutes and the car hasn't been ticketed yet
                if gap < 60 and not car["ticketed"]:
                    tickets += 1       # Give the car a ticket
                    car["ticketed"] = True  # Mark the car as ticketed
            
            # Update the car status to show it is now parked
            car["currently_parked"] = True
            car["enter_time"] = t  # Record the time the car entered
        
        else:
            # The car is currently parked and is now leaving the parking lot
            
            # Calculate how long the car stayed in the parking lot
            stay_duration = t - car["enter_time"]
            
            # If the stay was longer than 60 minutes and the car hasn't been ticketed yet
            if stay_duration > 60 and not car["ticketed"]:
                tickets += 1       # Give the car a ticket
                car["ticketed"] = True  # Mark the car as ticketed
            
            # Update the car status to show it has left the parking lot
            car["currently_parked"] = False
            car["last_exit"] = t  # Record the time the car left
    
    # After checking all events, return the total number of tickets issued
    return tickets


if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    n = int(input().strip())

    registrations = input().rstrip().split()

    times = list(map(int, input().rstrip().split()))

    tickets = ParkingTickets(n, registrations, times)

    fptr.write(str(tickets) + '\n')

    fptr.close()
```
