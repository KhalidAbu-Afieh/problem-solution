```python
#!/bin/python3

import math
import os
import random
import re
import sys

class BinarySearchTreeNode:
    def __init__(self, node_data):
        self.data = node_data
        self.left = None
        self.right = None

def insert_node_into_binary_search_tree(node, node_data):
    if node == None:
        node = BinarySearchTreeNode(node_data)
    else:
        if (node_data <= node.data):
            node.left = insert_node_into_binary_search_tree(node.left, node_data)
        else:
            node.right = insert_node_into_binary_search_tree(node.right, node_data)

    return node

def print_binary_search_tree_inorder_traversal(node, sep, fptr):
    if node == None:
        return

    print_binary_search_tree_inorder_traversal(node.left, sep, fptr)

    if node.left:
        fptr.write(sep)
    fptr.write(str(node.data))

    if node.right:
        fptr.write(sep)

    print_binary_search_tree_inorder_traversal(node.right, sep, fptr)

#
# Complete the 'GetDepth' function below.
#
# The function is expected to return an INTEGER.
# The function accepts following parameters:
#  1. INTEGER_BINARY_SEARCH_TREE tree
#  2. INTEGER k
#

#
# For your reference:
#
# BinarySearchTreeNode:
#     int data
#     BinarySearchTreeNode left
#     BinarySearchTreeNode right
#
#
def GetDepth(tree, k):
    depth = 0                 # start depth from 0
    current = tree            # start checking from the root node

    while current is not None:  # keep going while node exists
        if current.data == k:   # if current node value equals k
            return depth        # return the current depth

        elif k < current.data:  # if k is smaller than current value
            current = current.left  # move to the left child (a node that comes directly below another node)

        else:                   # if k is greater than current value
            current = current.right # move to the right child (a node that comes directly below another node)

        depth += 1              # increase depth after moving down

    return -1                   # return -1 if k is not found in the tree

if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    k = int(input().strip())

    tree_count = int(input().strip())

    tree = None

    for _ in range(tree_count):
        tree_item = int(input().strip())
        tree = insert_node_into_binary_search_tree(tree, tree_item)

    depth = GetDepth(tree, k)

    fptr.write(str(depth) + '\n')

    fptr.close()
```

```python
#!/bin/python3

import math
import os
import random
import re
import sys

class BinarySearchTreeNode:
    def __init__(self, node_data):
        self.data = node_data
        self.left = None
        self.right = None

def insert_node_into_binary_search_tree(node, node_data):
    if node == None:
        node = BinarySearchTreeNode(node_data)
    else:
        if (node_data <= node.data):
            node.left = insert_node_into_binary_search_tree(node.left, node_data)
        else:
            node.right = insert_node_into_binary_search_tree(node.right, node_data)

    return node

def print_binary_search_tree_inorder_traversal(node, sep, fptr):
    if node == None:
        return

    print_binary_search_tree_inorder_traversal(node.left, sep, fptr)

    if node.left:
        fptr.write(sep)
    fptr.write(str(node.data))

    if node.right:
        fptr.write(sep)

    print_binary_search_tree_inorder_traversal(node.right, sep, fptr)

#
# Complete the 'GetParent' function below.
#
# The function is expected to return an INTEGER.
# The function accepts following parameters:
#  1. INTEGER_BINARY_SEARCH_TREE tree
#  2. INTEGER k
#

#
# For your reference:
#
# BinarySearchTreeNode:
#     int data
#     BinarySearchTreeNode left
#     BinarySearchTreeNode right
#
#

def GetParent(tree, k):
    # if tree is empty no parent exists
    if tree is None:
        return -1

    current = tree          # start from the root node
    parent = None           # this will store the parent node

    while current is not None:
        # if we found the node with value k
        if current.data == k:
            # if parent is None then k is the root
            if parent is None:
                return -1
            else:
                return parent.data  # return parent value

        # if k is smaller move to left child
        elif k < current.data:
            parent = current        # current becomes parent
            current = current.left # move left

        # if k is greater move to right child
        else:
            parent = current         # current becomes parent (the node just above current)
            current = current.right # move right

    return -1

if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    k = int(input().strip())

    tree_count = int(input().strip())

    tree = None

    for _ in range(tree_count):
        tree_item = int(input().strip())
        tree = insert_node_into_binary_search_tree(tree, tree_item)

    parent = GetParent(tree, k)

    fptr.write(str(parent) + '\n')

    fptr.close()
```

```python
#!/bin/python3

import math
import os
import random
import re
import sys

#
# Complete the 'HeapType' function below.
#
# The function is expected to return a STRING.
# The function accepts INTEGER_ARRAY arr as parameter.
#

def HeapType(arr):
    n = len(arr)             # total number of elements
    is_max = True            # assume it is a max heap
    is_min = True            # assume it is a min heap

    # loop through each parent node
    for i in range(n):
        left = 2 * i + 1     # index of left child
        right = 2 * i + 2    # index of right child

        # check max heap condition
        if left < n and arr[i] < arr[left]:
            is_max = False
        if right < n and arr[i] < arr[right]:
            is_max = False

        # check min heap condition
        if left < n and arr[i] > arr[left]:
            is_min = False
        if right < n and arr[i] > arr[right]:
            is_min = False

    # decide the result
    if is_max:
        return "Max"
    elif is_min:
        return "Min"
    else:
        return "Neither"
    
if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    n = int(input().strip())

    arr = list(map(int, input().rstrip().split()))

    result = HeapType(arr)

    fptr.write(result + '\n')

    fptr.close()
```

```python
#!/bin/python3

import math
import os
import random
import re
import sys

class BinarySearchTreeNode:
    def __init__(self, node_data):
        self.data = node_data
        self.left = None
        self.right = None

def insert_node_into_binary_search_tree(node, node_data):
    if node == None:
        node = BinarySearchTreeNode(node_data)
    else:
        if (node_data <= node.data):
            node.left = insert_node_into_binary_search_tree(node.left, node_data)
        else:
            node.right = insert_node_into_binary_search_tree(node.right, node_data)

    return node

def print_binary_search_tree_inorder_traversal(node, sep, fptr):
    if node == None:
        return

    print_binary_search_tree_inorder_traversal(node.left, sep, fptr)

    if node.left:
        fptr.write(sep)
    fptr.write(str(node.data))

    if node.right:
        fptr.write(sep)

    print_binary_search_tree_inorder_traversal(node.right, sep, fptr)

#
# Complete the 'MaxEvenNodeSum' function below.
#
# The function is expected to return an INTEGER.
# The function accepts INTEGER_BINARY_SEARCH_TREE tree as parameter.
#

#
# For your reference:
#
# BinarySearchTreeNode:
#     int data
#     BinarySearchTreeNode left
#     BinarySearchTreeNode right
#
#
def MaxEvenNodeSum(tree):

    # helper function to calculate even path sum from a node
    def dfs(node):
        # if node does not exist return 0
        if node is None:
            return 0

          # if node value is odd stop the path
        if (node.data // 2) * 2 != node.data:
            return 0

        # calculate even path sum from left child
        left_sum = dfs(node.left)

        # calculate even path sum from right child
        right_sum = dfs(node.right)

        # return current value + max of left or right path
        return node.data + max(left_sum, right_sum)

    # start DFS from the root and return the result
    return dfs(tree)

if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    tree_count = int(input().strip())

    tree = None

    for _ in range(tree_count):
        tree_item = int(input().strip())
        tree = insert_node_into_binary_search_tree(tree, tree_item)

    maxSum = MaxEvenNodeSum(tree)

    fptr.write(str(maxSum) + '\n')

    fptr.close()
```
