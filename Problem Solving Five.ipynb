```python
#!/bin/python3

import math
import os
import random
import re
import sys

#
# Complete the 'CatOrDog' function below.
#
# The function is expected to return a STRING.
# The function accepts STRING word as parameter.
#

def CatOrDog(word):       

    word = word.lower()    # convert all letters to lowercase so case doesn't matter
    
    dog_count = 0          # variable to count letters d, o, g
    cat_count = 0          # variable to count letters c, a, t
    
    for ch in word:        # go through each character in the word
        
        if ch in "dog":    # check if character is d or o or g
            dog_count += 1 # add 1 to dog counter
            
        elif ch in "cat":  # check if character is c or a or t
            cat_count += 1 # add 1 to cat counter
    
    # compare counts after loop finishes
    if dog_count > cat_count:   # if dog letters appear more
        return "DOG"            # return DOG
    
    elif cat_count > dog_count: # if cat letters appear more
        return "CAT"            # return CAT
    
    else:                       # if equal or none found
        return "NEITHER"        # return NEITHER
if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    word = input()

    result = CatOrDog(word)

    fptr.write(result + '\n')

    fptr.close()
```

```python
#!/bin/python3

import math
import os
import random
import re
import sys

#
# Complete the 'Tickets' function below.
#
# The function is expected to return an INTEGER.
# The function accepts following parameters:
#  1. INTEGER nLines
#  2. STRING_ARRAY data
#

def Tickets(nLines, data):    
    times = {}               # dictionary to store first time of each car
    tickets = 0              # counter for speeding tickets
    
    limit_seconds = (0.5 / 55) * 3600   # max time allowed to avoid ticket
    
    for line in data:        # go through each record
        
        parts = line.split() # split string into pieces
        
        reg = parts[0]       # car registration
        h = int(parts[1])    # hour
        m = int(parts[2])    # minute
        s = int(parts[3])    # second
        
        total_seconds = h*3600 + m*60 + s   # convert time to seconds
        
        if reg not in times:   # first time we see this car
            times[reg] = total_seconds
        
        else:                  # second time we see the car
            
            diff = total_seconds - times[reg]  # time taken
            
            if diff <= limit_seconds:  # if too fast
                tickets += 1           # give ticket
    
    return tickets   # return total tickets
if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    nLines = int(input().strip())

    data = []

    for _ in range(nLines):
        data_item = input()
        data.append(data_item)

    ticketCount = Tickets(nLines, data)

    fptr.write(str(ticketCount) + '\n')

    fptr.close()
```

```python
#!/bin/python3

import math
import os
import random
import re
import sys

#
# Complete the 'IsValidSudoku' function below.
#
# The function is expected to return a STRING.
# The function accepts STRING s as parameter.
#
def IsValidSudoku(s):   

    # make empty grid list
    grid = []
    
    # loop 9 times to make 9 rows
    for i in range(9):
        
        # take 9 characters from string
        row = list(s[i*9:(i+1)*9])
        
        # add row to grid
        grid.append(row)
    

    for r in range(9):   # go through each row
        
        # set removes duplicates
        if len(set(grid[r])) != 9:  # check if row does not contain 9 different numbers
            return "INVALID" # sudoku is wrong stop and say INVALID
        
    for c in range(9):   # go through each column
        
        col = []   # empty list for column
        
        for r in range(9):   # go through rows
            col.append(grid[r][c])   # take column value
        
        if len(set(col)) != 9: # if the column does not have 9 different numbers
            return "INVALID"
    
    for br in range(0, 9, 3):   # box row start (0,3,6)
        
        for bc in range(0, 9, 3):   # box column start (0,3,6)
            
            box = []   # store numbers in box
            
            for r in range(3):   # inside box rows
                
                for c in range(3):   # inside box columns
                    
                    box.append(grid[br+r][bc+c])
            
            if len(set(box)) != 9: # if 3x3 box does not have 9 different numbers
                return "INVALID"
    
    
    # if everything correct
    return "VALID"
if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    s = input()

    result = IsValidSudoku(s)

    fptr.write(result + '\n')

    fptr.close()
```

```python
#!/bin/python3

import math
import os
import random
import re
import sys

#
# Complete the 'ValidBrackets' function below.
#
# The function is expected to return a STRING.
# The function accepts STRING s as parameter.
#

def ValidBrackets(s):
    stack = []  # empty list to act as a stack
    
    # dictionary to match closing to opening brackets
    pairs = {')':'(', ']':'[', '}':'{'}
    
    for ch in s:  # go through each character in the string
        
        if ch in "([{":      # if it is an opening bracket
            stack.append(ch) # push it onto the stack
            
        elif ch in ")]}":    # if it is a closing bracket
            
            if not stack:    # stack empty but found closing invalid
                return "INVALID"
            
            if stack[-1] == pairs[ch]:  # top of stack matches
                stack.pop()             # remove it properly closed
            else:
                return "INVALID"        # top doesn't match invalid
                
    # after processing all brackets
    if stack:           # anything left in stack unclosed brackets
        return "INVALID"
    
    return "VALID"      # all brackets matched properly

if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    s = input()

    result = ValidBrackets(s)

    fptr.write(result + '\n')

    fptr.close()
```
